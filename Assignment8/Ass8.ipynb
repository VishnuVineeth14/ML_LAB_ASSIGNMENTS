```python
import pandas as pd
```

```python
X_train = pd.read_csv(
    "UCI HAR Dataset/train/X_train.txt",
    sep=r"\s+",
    header=None
)

X_test = pd.read_csv(
    "UCI HAR Dataset/test/X_test.txt",
    sep=r"\s+",
    header=None
)
```

```python
X=pd.concat([X_train,X_test],ignore_index=True)

print(X.shape)
```

```python
y_train = pd.read_csv(
    "UCI HAR Dataset/train/y_train.txt",
    sep=r"\s+",
    header=None
)

y_test = pd.read_csv(
    "UCI HAR Dataset/test/y_test.txt",
    sep=r"\s+",
    header=None
)

y = pd.concat([y_train, y_test], ignore_index=True)
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("Combined X shape:", X.shape)

print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
print("Combined y shape:", y.shape)
```

```python


print(f"Null values of x: {X.isnull().sum().sum()}")
print(f"Null values of y: {y.isnull().sum().sum()}")


print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFirst 5 rows of X:")
print(X.head())

print("\nFirst 5 rows of y:")
print(y.head())

print("\nX information:")
X.info()

```

```python
y.columns = ["activity"]

# Load activity labels
activity_labels = pd.read_csv(
    "UCI HAR Dataset/activity_labels.txt",
    sep=r"\s+",
    header=None,
    names=["activity_id", "activity_name"]
)

# Give the y column a meaningful name
y.columns = ["activity"]

print(activity_labels)

print("\nActivity counts:")
print(y["activity"].value_counts().sort_index())
```

```python
import matplotlib.pyplot as plt

y["activity"].value_counts().sort_index().plot(kind="bar")

plt.xlabel("Activity")
plt.ylabel("Number of Samples")
plt.title("Distribution of Human Activities")
plt.show()
```
